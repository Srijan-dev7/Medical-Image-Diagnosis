```python
import os
import matplotlib.pyplot as plt
from PIL import Image
```

```python
data_path="../data/chest_xray/"
train_path=os.path.join(data_path,"train")
test_path=os.path.join(data_path,"test")
val_path=os.path.join(data_path,"val")

```

```python
print("Train:", train_path)
print("Test:", test_path)
print("Validation:", val_path)
print("Training classes:", os.listdir(train_path))
print("Testing classes:", os.listdir(test_path))
print("Validation classes:", os.listdir(val_path))
```

```python
classes=["NORMAL","PNEUMONIA"]
for cls in classes:
    train_count=len(os.listdir(os.path.join(train_path,cls)))
    test_count=len(os.listdir(os.path.join(test_path,cls)))
    val_count=len(os.listdir(os.path.join(val_path,cls)))

    print(cls)
    print("Train",train_count)
    print("Test",test_count)
    print("Validation",val_count)
```

```python
train_counts=[]
for cls in classes:
    count=len(os.listdir(os.path.join(train_path,cls)))
    train_counts.append(count)

plt.bar(classes, train_counts)
plt.xlabel("Class")
plt.ylabel("Number of Images")
plt.title("Training Dataset Class Distribution")
plt.show()

```

```python
normal_files = [
    file for file in os.listdir(os.path.join(train_path, "NORMAL"))
    if file.lower().endswith((".jpeg", ".jpg", ".png"))
]

pneumonia_files = [
    file for file in os.listdir(os.path.join(train_path, "PNEUMONIA"))
    if file.lower().endswith((".jpeg", ".jpg", ".png"))
]

print("Normal image:", normal_files[0])
print("Pneumonia image:", pneumonia_files[0])
```

```python
normal_img = Image.open(
    os.path.join(train_path, "NORMAL", normal_files[0])
)

pneumonia_img = Image.open(
    os.path.join(train_path, "PNEUMONIA", pneumonia_files[0])
)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(normal_img, cmap="gray")
plt.title("Normal")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(pneumonia_img, cmap="gray")
plt.title("Pneumonia")
plt.axis("off")

plt.show()
```
